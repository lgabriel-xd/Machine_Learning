```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

# Preparar datos
X = df[['latitude', 'longitude', 'brightness']].values
X_scaled = StandardScaler().fit_transform(X)

# Calcular inercia para diferentes números de clusters
inertias = []
k_range = range(1, 15)

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    inertias.append(kmeans.inertia_)

# Graficar método del codo
plt.figure(figsize=(10, 6))
plt.plot(k_range, inertias, 'bo-')
plt.xlabel('Número de Clusters')
plt.ylabel('Inercia')
plt.title('Método del Codo para K Óptimo')
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Cargar datos
df = pd.read_csv('dataset_incendios_2000_muestras.csv')

# 1. Preparar datos - solo coordenadas
X = df[['latitude', 'longitude']].values

# Crear y guardar el scaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 2. Aplicar KMeans con 5 clusters
kmeans = KMeans(n_clusters=5, random_state=42)
df['cluster'] = kmeans.fit_predict(X_scaled)

# 3. Convertir centroides escalados a coordenadas originales
centers_original = scaler.inverse_transform(kmeans.cluster_centers_)

# Mostrar información
print("Distribución de puntos por cluster:")
print(df['cluster'].value_counts().sort_index())
print(f"\nCoordenadas de los centroides:")
for i, center in enumerate(centers_original):
    print(f"Cluster {i}: Lat {center[0]:.4f}, Lon {center[1]:.4f}")
```

```python
# Visualización básica CORREGIDA
plt.figure(figsize=(14, 10))

# Scatter plot de clusters
scatter = plt.scatter(df['longitude'], df['latitude'], 
                     c=df['cluster'], cmap='viridis', 
                     alpha=0.7, s=40, edgecolors='white', linewidth=0.3)

# Marcar centroides (USANDO centers_original que YA está definido)
plt.scatter(centers_original[:, 1], centers_original[:, 0], 
           marker='X', s=200, c='red', edgecolors='black', linewidth=2,
           label='Centroides')

plt.colorbar(scatter, label='Cluster')
plt.xlabel('Longitud', fontsize=12)
plt.ylabel('Latitud', fontsize=12)
plt.title('Clustering de Incendios - Solo Coordenadas Geográficas\n(5 Clusters KMeans)', fontsize=14)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Colores más distintivos
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']

plt.figure(figsize=(14, 10))

for cluster_id in range(5):
    cluster_data = df[df['cluster'] == cluster_id]
    plt.scatter(cluster_data['longitude'], cluster_data['latitude'], 
               c=colors[cluster_id], label=f'Cluster {cluster_id}', 
               alpha=0.7, s=40, edgecolors='white', linewidth=0.3)

# Centroides (USANDO centers_original que YA está definido)
plt.scatter(centers_original[:, 1], centers_original[:, 0], 
           marker='X', s=300, c='black', edgecolors='white', linewidth=2,
           label='Centroides', zorder=5)

plt.xlabel('Longitud', fontsize=12)
plt.ylabel('Latitud', fontsize=12)
plt.title('Clusters de Incendios por Zona Geográfica', fontsize=14)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Cargar y preparar datos
df = pd.read_csv('dataset_incendios_2000_muestras.csv')
X = df[['latitude', 'longitude']].values

# Escalar y clusterizar
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
kmeans = KMeans(n_clusters=5, random_state=42)
df['cluster'] = kmeans.fit_predict(X_scaled)
centers_original = scaler.inverse_transform(kmeans.cluster_centers_)

# Visualización completa
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 8))

# Gráfico 1: Básico
sc1 = ax1.scatter(df['longitude'], df['latitude'], 
                 c=df['cluster'], cmap='viridis', alpha=0.7, s=40)
ax1.scatter(centers_original[:, 1], centers_original[:, 0], 
           marker='X', s=200, c='red', edgecolors='black', linewidth=2)
ax1.set_xlabel('Longitud')
ax1.set_ylabel('Latitud')
ax1.set_title('Clusters con Centroides')
ax1.grid(True, alpha=0.3)

# Gráfico 2: Por colores
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']
for cluster_id in range(5):
    cluster_data = df[df['cluster'] == cluster_id]
    ax2.scatter(cluster_data['longitude'], cluster_data['latitude'], 
               c=colors[cluster_id], label=f'Cluster {cluster_id}', 
               alpha=0.7, s=40)
ax2.scatter(centers_original[:, 1], centers_original[:, 0], 
           marker='X', s=200, c='black', edgecolors='white', linewidth=2)
ax2.set_xlabel('Longitud')
ax2.set_ylabel('Latitud')
ax2.set_title('Clusters por Color')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Resumen
print("=== RESUMEN ===")
print("Puntos por cluster:")
print(df['cluster'].value_counts().sort_index())
```
